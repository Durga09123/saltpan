```python
# ============================================================
# TEST CANDIDATE SCORING RANGES
# ============================================================

# Candidate thresholds based on the observed daily distributions

def temperature_score(x):
    if 30 <= x < 34:
        return 4
    elif 28 <= x < 30:
        return 3
    elif 26 <= x < 28:
        return 2
    else:
        return 1


def humidity_score(x):
    if 50 <= x < 65:
        return 4
    elif 65 <= x < 75:
        return 3
    elif 75 <= x < 85:
        return 2
    else:
        return 1


def wind_score(x):
    if 5 <= x < 8:
        return 4
    elif 4 <= x < 5:
        return 3
    elif 3 <= x < 4:
        return 2
    else:
        return 1


def rainfall_score(x):
    if x <= 0.5:
        return 4
    elif x <= 2:
        return 3
    elif x <= 5:
        return 2
    else:
        return 1


# Apply candidate scores
daily["T_candidate"] = daily["temperature_2m (°C)"].apply(temperature_score)
daily["H_candidate"] = daily["relative_humidity_2m (%)"].apply(humidity_score)
daily["W_candidate"] = daily["wind_speed_10m (m/s)"].apply(wind_score)
daily["R_candidate"] = daily["precipitation (mm)"].apply(rainfall_score)


# Display score distributions
for column in [
    "T_candidate",
    "H_candidate",
    "W_candidate",
    "R_candidate"
]:
    print("\n", column)
    print(
        daily[column]
        .value_counts()
        .sort_index()
        .rename_axis("Score")
        .to_frame("Number_of_Days")
    )

    print(
        (
            daily[column]
            .value_counts(normalize=True)
            .sort_index() * 100
        ).round(2)
        .rename("Percentage")
    )
```

```python
# ============================================================
# TEST CANDIDATE OVERALL SUITABILITY SCORE
# ============================================================

daily["Candidate_Total_Score"] = (
    daily["T_candidate"]
    + daily["H_candidate"]
    + daily["W_candidate"]
    + daily["R_candidate"]
)

daily["Candidate_SI"] = (
    daily["Candidate_Total_Score"] / 16 * 100
)

print("Candidate Suitability Index distribution:")
print(
    daily["Candidate_SI"]
    .describe()
    .round(2)
)

print("\nSuitability Index ranges:")

si_bins = pd.cut(
    daily["Candidate_SI"],
    bins=[0, 25, 50, 62.5, 75, 87.5, 100],
    labels=[
        "0–25%",
        ">25–50%",
        ">50–62.5%",
        ">62.5–75%",
        ">75–87.5%",
        ">87.5–100%"
    ],
    include_lowest=True
)

si_distribution = (
    si_bins.value_counts()
    .sort_index()
)

si_percentage = (
    si_distribution / len(daily) * 100
).round(2)

si_summary = pd.DataFrame({
    "Number_of_Days": si_distribution,
    "Percentage": si_percentage
})

print(si_summary)
```

```python
# ============================================================
# WEATHER CONDITIONS AT DIFFERENT SUITABILITY LEVELS
# ============================================================

daily["SI_Category"] = pd.cut(
    daily["Candidate_SI"],
    bins=[0, 50, 62.5, 75, 87.5, 100],
    labels=[
        "0–50%",
        ">50–62.5%",
        ">62.5–75%",
        ">75–87.5%",
        ">87.5–100%"
    ],
    include_lowest=True
)

si_weather_summary = (
    daily.groupby("SI_Category", observed=True)[
        [
            "temperature_2m (°C)",
            "relative_humidity_2m (%)",
            "precipitation (mm)",
            "wind_speed_10m (m/s)"
        ]
    ]
    .agg(["mean", "median", "min", "max", "count"])
    .round(2)
)

print(si_weather_summary)
```

```python
# ============================================================
# CHECK EACH WEATHER FACTOR'S CONTRIBUTION TO TOTAL SCORE
# ============================================================

score_contribution = pd.DataFrame({
    "Temperature_Score": daily["T_candidate"],
    "Humidity_Score": daily["H_candidate"],
    "Wind_Score": daily["W_candidate"],
    "Rainfall_Score": daily["R_candidate"]
})

print("Score correlation:")
print(
    score_contribution
    .corr()
    .round(3)
)

print("\nAverage score by SI category:")

daily["T_Score"] = daily["T_candidate"]
daily["H_Score"] = daily["H_candidate"]
daily["W_Score"] = daily["W_candidate"]
daily["R_Score"] = daily["R_candidate"]

factor_by_category = (
    daily.groupby("SI_Category", observed=True)[
        ["T_Score", "H_Score", "W_Score", "R_Score"]
    ]
    .mean()
    .round(2)
)

print(factor_by_category)
```

```python
# ============================================================
# RAINFALL RANGE ANALYSIS
# ============================================================

rain_bins = pd.cut(
    daily["precipitation (mm)"],
    bins=[-0.001, 0, 0.5, 1, 2, 5, 10, 20, float("inf")],
    labels=[
        "0 mm",
        ">0–0.5 mm",
        ">0.5–1 mm",
        ">1–2 mm",
        ">2–5 mm",
        ">5–10 mm",
        ">10–20 mm",
        ">20 mm"
    ],
    include_lowest=True
)

rain_analysis = (
    daily.groupby(rain_bins, observed=True)
    .agg(
        Number_of_Days=("Candidate_SI", "count"),
        Mean_Temperature=("temperature_2m (°C)", "mean"),
        Mean_Humidity=("relative_humidity_2m (%)", "mean"),
        Mean_Rainfall=("precipitation (mm)", "mean"),
        Mean_Wind=("wind_speed_10m (m/s)", "mean"),
        Mean_SI=("Candidate_SI", "mean")
    )
    .round(2)
)

rain_analysis["Percentage"] = (
    rain_analysis["Number_of_Days"]
    / len(daily)
    * 100
).round(2)

print(rain_analysis)
```

```python
# ============================================================
# COMBINED WEATHER CONDITION ANALYSIS
# ============================================================

# Create percentile-based categories for each weather parameter.
# These are used only for exploratory analysis, NOT as final labels.

daily["Temp_Percentile"] = pd.qcut(
    daily["temperature_2m (°C)"],
    q=4,
    labels=["Low", "Moderate", "High", "Very High"]
)

daily["Humidity_Percentile"] = pd.qcut(
    daily["relative_humidity_2m (%)"],
    q=4,
    labels=["Low", "Moderate", "High", "Very High"]
)

daily["Wind_Percentile"] = pd.qcut(
    daily["wind_speed_10m (m/s)"],
    q=4,
    labels=["Low", "Moderate", "High", "Very High"]
)

daily["Rain_Percentile"] = pd.qcut(
    daily["precipitation (mm)"],
    q=4,
    duplicates="drop"
)

print("Temperature categories:")
print(daily["Temp_Percentile"].value_counts().sort_index())

print("\nHumidity categories:")
print(daily["Humidity_Percentile"].value_counts().sort_index())

print("\nWind categories:")
print(daily["Wind_Percentile"].value_counts().sort_index())

print("\nRainfall percentile categories:")
print(daily["Rain_Percentile"].value_counts().sort_index())
```

```python
# ============================================================
# EXPLORATORY PHYSICAL SUITABILITY SCORE
# ============================================================

# Temperature
def temp_physical_score(x):
    if x >= 32:
        return 4
    elif x >= 29:
        return 3
    elif x >= 26:
        return 2
    else:
        return 1


# Humidity
def humidity_physical_score(x):
    if x < 50:
        return 4
    elif x < 65:
        return 3
    elif x < 75:
        return 2
    else:
        return 1


# Wind
def wind_physical_score(x):
    if 5 <= x < 8:
        return 4
    elif 8 <= x < 10:
        return 3
    elif 3 <= x < 5:
        return 2
    else:
        return 1


# Rainfall
def rain_physical_score(x):
    if x <= 0.5:
        return 4
    elif x <= 2:
        return 3
    elif x <= 5:
        return 2
    else:
        return 1


# Apply scores
daily["T_Physical"] = daily["temperature_2m (°C)"].apply(
    temp_physical_score
)

daily["H_Physical"] = daily["relative_humidity_2m (%)"].apply(
    humidity_physical_score
)

daily["W_Physical"] = daily["wind_speed_10m (m/s)"].apply(
    wind_physical_score
)

daily["R_Physical"] = daily["precipitation (mm)"].apply(
    rain_physical_score
)


# Total exploratory score
daily["Physical_Total"] = (
    daily["T_Physical"]
    + daily["H_Physical"]
    + daily["W_Physical"]
    + daily["R_Physical"]
)


# Suitability index
daily["Physical_SI"] = (
    daily["Physical_Total"] / 16 * 100
)


print("Physical SI statistics:")
print(daily["Physical_SI"].describe().round(2))


print("\nPhysical SI distribution:")
print(
    pd.cut(
        daily["Physical_SI"],
        bins=[0, 50, 62.5, 75, 87.5, 100],
        include_lowest=True
    ).value_counts().sort_index()
)


print("\nScore combination counts:")
print(
    daily[
        [
            "T_Physical",
            "H_Physical",
            "W_Physical",
            "R_Physical"
        ]
    ]
    .value_counts()
    .head(20)
)
```

```python
# ============================================================
# HIGH AND LOW COMBINED WEATHER CONDITIONS
# ============================================================

# Select days where all four exploratory scores are favorable
high_condition = daily[
    (daily["T_Physical"] >= 3) &
    (daily["H_Physical"] >= 3) &
    (daily["W_Physical"] >= 3) &
    (daily["R_Physical"] >= 3)
]

# Select days where all four exploratory scores are unfavorable
low_condition = daily[
    (daily["T_Physical"] <= 2) &
    (daily["H_Physical"] <= 2) &
    (daily["W_Physical"] <= 2) &
    (daily["R_Physical"] <= 2)
]


print("============================================================")
print("HIGH-CONDITION DAYS")
print("============================================================")

print("Number of days:", len(high_condition))

print(
    high_condition[
        [
            "temperature_2m (°C)",
            "relative_humidity_2m (%)",
            "wind_speed_10m (m/s)",
            "precipitation (mm)"
        ]
    ].describe().round(2)
)


print("\n============================================================")
print("LOW-CONDITION DAYS")
print("============================================================")

print("Number of days:", len(low_condition))

print(
    low_condition[
        [
            "temperature_2m (°C)",
            "relative_humidity_2m (%)",
            "wind_speed_10m (m/s)",
            "precipitation (mm)"
        ]
    ].describe().round(2)
)
```

```python
# ============================================================
# COMPARE HIGH vs LOW CONDITIONS AT EACH PARAMETER
# ============================================================

parameters = {
    "Temperature": "temperature_2m (°C)",
    "Humidity": "relative_humidity_2m (%)",
    "Wind": "wind_speed_10m (m/s)",
    "Rainfall": "precipitation (mm)"
}

for name, column in parameters.items():

    print("\n" + "=" * 65)
    print(name)
    print("=" * 65)

    comparison = pd.DataFrame({
        "High_Condition": high_condition[column].quantile(
            [0.05, 0.25, 0.50, 0.75, 0.95]
        ),
        "Low_Condition": low_condition[column].quantile(
            [0.05, 0.25, 0.50, 0.75, 0.95]
        )
    })

    comparison.index = ["5%", "25%", "50%", "75%", "95%"]

    print(comparison.round(2))
```

```python
# ============================================================
# FINAL CANDIDATE SCORING SYSTEM - VERSION 2
# ============================================================

def temperature_score_v2(x):
    if x >= 32:
        return 4
    elif x >= 29:
        return 3
    elif x >= 26:
        return 2
    else:
        return 1


def humidity_score_v2(x):
    if x < 50:
        return 4
    elif x < 65:
        return 3
    elif x < 75:
        return 2
    else:
        return 1


def wind_score_v2(x):
    if 5 <= x < 8:
        return 4
    elif 8 <= x < 10:
        return 3
    elif 3 <= x < 5:
        return 2
    else:
        return 1


def rainfall_score_v2(x):
    if x <= 0.5:
        return 4
    elif x <= 2:
        return 3
    elif x <= 5:
        return 2
    else:
        return 1


# Apply scores
daily["T_Score_v2"] = daily[
    "temperature_2m (°C)"
].apply(temperature_score_v2)

daily["H_Score_v2"] = daily[
    "relative_humidity_2m (%)"
].apply(humidity_score_v2)

daily["W_Score_v2"] = daily[
    "wind_speed_10m (m/s)"
].apply(wind_score_v2)

daily["R_Score_v2"] = daily[
    "precipitation (mm)"
].apply(rainfall_score_v2)


# Total score
daily["Total_Score_v2"] = (
    daily["T_Score_v2"]
    + daily["H_Score_v2"]
    + daily["W_Score_v2"]
    + daily["R_Score_v2"]
)


# Suitability Index
daily["SI_v2"] = (
    daily["Total_Score_v2"] / 16 * 100
)


print("============================================================")
print("V2 SUITABILITY INDEX STATISTICS")
print("============================================================")

print(
    daily["SI_v2"]
    .describe()
    .round(2)
)


print("\n============================================================")
print("SCORE DISTRIBUTION")
print("============================================================")

print(
    daily["Total_Score_v2"]
    .value_counts()
    .sort_index()
)


print("\n============================================================")
print("SUITABILITY INDEX DISTRIBUTION")
print("============================================================")

print(
    pd.cut(
        daily["SI_v2"],
        bins=[0, 50, 62.5, 75, 87.5, 100],
        include_lowest=True
    )
    .value_counts()
    .sort_index()
)


print("\n============================================================")
print("AVERAGE WEATHER BY SI")
print("============================================================")

si_analysis = (
    daily.groupby(
        pd.cut(
            daily["SI_v2"],
            bins=[0, 50, 62.5, 75, 87.5, 100],
            include_lowest=True
        ),
        observed=True
    )
    .agg(
        Number_of_Days=("SI_v2", "count"),
        Temperature=("temperature_2m (°C)", "mean"),
        Humidity=("relative_humidity_2m (%)", "mean"),
        Wind=("wind_speed_10m (m/s)", "mean"),
        Rainfall=("precipitation (mm)", "mean")
    )
    .round(2)
)

print(si_analysis)
```

```python
# ============================================================
# TEST SCORE >= 12 AS THE SUITABILITY THRESHOLD
# ============================================================

daily["Suitability_v2"] = (
    daily["Total_Score_v2"] >= 12
).astype(int)

daily["Suitability_Label_v2"] = daily[
    "Suitability_v2"
].map({
    1: "Suitable",
    0: "Not Suitable"
})


print("============================================================")
print("CLASS DISTRIBUTION")
print("============================================================")

class_counts = daily["Suitability_Label_v2"].value_counts()

print(class_counts)

print("\nClass percentages:")

class_percentages = (
    daily["Suitability_Label_v2"]
    .value_counts(normalize=True)
    .mul(100)
    .round(2)
)

print(class_percentages)


print("\n============================================================")
print("AVERAGE WEATHER BY CLASS")
print("============================================================")

class_analysis = (
    daily.groupby("Suitability_Label_v2")
    .agg(
        Number_of_Days=("Suitability_v2", "count"),
        Mean_Temperature=("temperature_2m (°C)", "mean"),
        Mean_Humidity=("relative_humidity_2m (%)", "mean"),
        Mean_Wind=("wind_speed_10m (m/s)", "mean"),
        Mean_Rainfall=("precipitation (mm)", "mean"),
        Mean_Total_Score=("Total_Score_v2", "mean"),
        Mean_SI=("SI_v2", "mean")
    )
    .round(2)
)

print(class_analysis)


print("\n============================================================")
print("SCORE-WISE CLASSIFICATION")
print("============================================================")

score_class = (
    daily.groupby("Total_Score_v2")
    .agg(
        Number_of_Days=("Suitability_v2", "count"),
        Suitable_Days=("Suitability_v2", "sum")
    )
)

score_class["Not_Suitable_Days"] = (
    score_class["Number_of_Days"]
    - score_class["Suitable_Days"]
)

score_class["Suitable_Percentage"] = (
    score_class["Suitable_Days"]
    / score_class["Number_of_Days"]
    * 100
).round(2)

print(score_class)
```

```python
# ============================================================
# FINAL LABEL VALIDATION
# ============================================================

print("Dataset shape:")
print(daily.shape)

print("\nColumns:")
print(daily.columns.tolist())

print("\nMissing values in final scoring columns:")
print(
    daily[
        [
            "T_Score_v2",
            "H_Score_v2",
            "W_Score_v2",
            "R_Score_v2",
            "Total_Score_v2",
            "SI_v2",
            "Suitability_v2",
            "Suitability_Label_v2"
        ]
    ].isnull().sum()
)

print("\nDuplicate rows:")
print(daily.duplicated().sum())

print("\nTotal score range:")
print(
    daily["Total_Score_v2"].min(),
    "to",
    daily["Total_Score_v2"].max()
)

print("\nSI range:")
print(
    daily["SI_v2"].min(),
    "to",
    daily["SI_v2"].max()
)

print("\nFinal labels:")
print(
    daily["Suitability_Label_v2"].value_counts()
)

print("\nSample final records:")
print(
    daily[
        [
            "location_id",
            "Date",
            "temperature_2m (°C)",
            "relative_humidity_2m (%)",
            "wind_speed_10m (m/s)",
            "precipitation (mm)",
            "T_Score_v2",
            "H_Score_v2",
            "W_Score_v2",
            "R_Score_v2",
            "Total_Score_v2",
            "SI_v2",
            "Suitability_Label_v2"
        ]
    ].head(10)
)
```

```python
# ============================================================
# FINAL DATE-BASED TRAIN-TEST SPLIT
# ============================================================

# Sort by date and location
ml_data_sorted = ml_data.sort_values(
    by=["Date", "location_id"]
).reset_index(drop=True)


# Get unique dates in chronological order
unique_dates = ml_data_sorted["Date"].drop_duplicates().sort_values().reset_index(drop=True)


# 80% of unique dates for training
date_split_index = int(len(unique_dates) * 0.80)

train_end_date = unique_dates.iloc[date_split_index - 1]
test_start_date = unique_dates.iloc[date_split_index]


# Create masks
train_mask = ml_data_sorted["Date"] <= train_end_date
test_mask = ml_data_sorted["Date"] >= test_start_date


# Create X and y
X_train = ml_data_sorted.loc[train_mask, feature_columns].copy()
X_test = ml_data_sorted.loc[test_mask, feature_columns].copy()

y_train = ml_data_sorted.loc[train_mask, target_column].copy()
y_test = ml_data_sorted.loc[test_mask, target_column].copy()


print("============================================================")
print("FINAL DATE-BASED TRAIN-TEST SPLIT")
print("============================================================")

print("Total unique dates:", len(unique_dates))
print("Training unique dates:", train_mask.sum() // 5)
print("Testing unique dates:", test_mask.sum() // 5)

print("\nTraining features:", X_train.shape)
print("Testing features :", X_test.shape)

print("\nTraining target:", y_train.shape)
print("Testing target :", y_test.shape)


print("\n============================================================")
print("CLASS DISTRIBUTION")
print("============================================================")

print("\nTraining:")
print(y_train.value_counts())

print("\nTesting:")
print(y_test.value_counts())


print("\n============================================================")
print("DATE PERIOD")
print("============================================================")

print("\nTraining period:")
print(ml_data_sorted.loc[train_mask, "Date"].min())
print("to")
print(ml_data_sorted.loc[train_mask, "Date"].max())

print("\nTesting period:")
print(ml_data_sorted.loc[test_mask, "Date"].min())
print("to")
print(ml_data_sorted.loc[test_mask, "Date"].max())


print("\n============================================================")
print("DATE OVERLAP CHECK")
print("============================================================")

train_dates = set(ml_data_sorted.loc[train_mask, "Date"])
test_dates = set(ml_data_sorted.loc[test_mask, "Date"])

overlap = train_dates.intersection(test_dates)

print("Overlapping dates:", len(overlap))

if len(overlap) == 0:
    print("✓ No date overlap")
else:
    print("✗ Date overlap found:", sorted(overlap)[:10])
```

```python
# ============================================================
# TRAIN RANDOM FOREST MODEL
# ============================================================

from sklearn.ensemble import RandomForestClassifier


# Create Random Forest model
rf_model = RandomForestClassifier(
    n_estimators=200,
    random_state=42,
    class_weight="balanced",
    n_jobs=-1
)


print("============================================================")
print("TRAINING RANDOM FOREST")
print("============================================================")

rf_model.fit(X_train, y_train)

print("Random Forest training completed successfully.")


print("\n============================================================")
print("MODEL INFORMATION")
print("============================================================")

print("Number of trees:", rf_model.n_estimators)
print("Number of features:", rf_model.n_features_in_)
print("Training samples:", X_train.shape[0])
print("Testing samples:", X_test.shape[0])


print("\nFeatures used:")
for feature in feature_columns:
    print("-", feature)
```

```python
# ============================================================
# EVALUATE RANDOM FOREST MODEL
# ============================================================

from sklearn.metrics import (
    accuracy_score,
    precision_score,
    recall_score,
    f1_score,
    confusion_matrix,
    classification_report
)


# Predict test data
y_pred = rf_model.predict(X_test)


# Calculate metrics
accuracy = accuracy_score(y_test, y_pred)
precision = precision_score(y_test, y_pred, zero_division=0)
recall = recall_score(y_test, y_pred, zero_division=0)
f1 = f1_score(y_test, y_pred, zero_division=0)

cm = confusion_matrix(y_test, y_pred)


print("============================================================")
print("RANDOM FOREST TEST RESULTS")
print("============================================================")

print(f"Accuracy : {accuracy * 100:.2f}%")
print(f"Precision: {precision * 100:.2f}%")
print(f"Recall   : {recall * 100:.2f}%")
print(f"F1 Score : {f1 * 100:.2f}%")


print("\n============================================================")
print("CONFUSION MATRIX")
print("============================================================")

print(cm)


print("\n============================================================")
print("CLASSIFICATION REPORT")
print("============================================================")

print(
    classification_report(
        y_test,
        y_pred,
        target_names=["Not Suitable", "Suitable"],
        zero_division=0
    )
)
```

```python
# ============================================================
# TEST SET PREDICTION PROBABILITY ANALYSIS
# ============================================================

test_probabilities = rf_model.predict_proba(X_test)

test_suitable_probability = test_probabilities[:, 1]


print("============================================================")
print("SUITABLE PROBABILITY ANALYSIS")
print("============================================================")

print(
    "Minimum probability:",
    round(test_suitable_probability.min() * 100, 2),
    "%"
)

print(
    "Maximum probability:",
    round(test_suitable_probability.max() * 100, 2),
    "%"
)

print(
    "Mean probability:",
    round(test_suitable_probability.mean() * 100, 2),
    "%"
)

print(
    "Median probability:",
    round(pd.Series(test_suitable_probability).median() * 100, 2),
    "%"
)


print("\n============================================================")
print("PROBABILITY RANGES")
print("============================================================")

probability_ranges = pd.cut(
    test_suitable_probability,
    bins=[0, 0.25, 0.50, 0.75, 1.00],
    labels=[
        "0–25%",
        "25–50%",
        "50–75%",
        "75–100%"
    ],
    include_lowest=True
)

print(
    probability_ranges.value_counts()
    .sort_index()
)


print("\n============================================================")
print("PREDICTION VS ACTUAL")
print("============================================================")

comparison = pd.DataFrame({
    "Actual": y_test.values,
    "Predicted": y_pred,
    "Suitable_Probability": test_suitable_probability
})

print(comparison.head(10))
```
